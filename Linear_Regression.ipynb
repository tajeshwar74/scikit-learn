```python
import pandas as pd
```

```python
import numpy as np
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
from sklearn.linear_model import LinearRegression
```

```python
np.random.seed(42)

num_samples = 500

years_of_experience = np.random.randint(2, 21, size=num_samples)
slope = (200_000 - 60_000) / 18
intercept = 60_000

salaries = slope * years_of_experience + intercept + np.random.normal(0, 10_000, size=num_samples)

data = {'Years_of_Experience': years_of_experience, 'Salary': salaries}

df = pd.DataFrame(data)
```

```python
df.describe()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='Years_of_Experience', y='Salary', data=df, color='red')
sns.regplot(x='Years_of_Experience', y='Salary', data=df, scatter=False, color='blue')
```

```python
X = df[['Years_of_Experience']]
```

```python
y = df['Salary']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=13)
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
lr.score(X_train, y_train)
```

```python
lr.score(X_test, y_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
y_pred = lr.predict(X_test)
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
mean_squared_error(y_test, y_pred)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
coefficients = lr.coef_
intercept = lr.intercept_

x = np.linspace(0,20,100)
y = coefficients * x + intercept

plt.scatter(x, y, label=f'y={coefficients[0]}x + {intercept}', color='blue')

plt.xlabel('years of experience')
plt.ylabel('salary')

plt.grid()
plt.show()
```
