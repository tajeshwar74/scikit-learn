```python
import numpy as np
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

# Define lists for categorical data
continents = ['North America', 'Europe', 'Asia', 'South America', 'Australia', 'Africa']
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# Generate 42 rows of synthetic data
data = {
    'City_population': np.random.randint(100000, 10000000, size=42),
    'Continent': np.random.choice(continents, size=42),
    'Venue_capacity': np.random.randint(500, 80000, size=42),
    'Day_of_week': np.random.choice(days, size=42),
    'Multiple_concerts': np.random.choice([0, 1], size=42),
    'sold_out': np.random.choice([0, 1], size=42)
}

df = pd.DataFrame(data)
df.head()
```

```python
df2 = pd.get_dummies(df[['Continent', 'Day_of_week']], dtype=int)
```

```python
df2.head()
```

```python
df3 = pd.concat([df, df2], axis = 1)
```

```python
df3.head()
```

```python
df4 = df3.drop(['Continent', 'Day_of_week'], axis=1)
```

```python
df4.head()
```

```python
X = df4.drop(['sold_out'], axis=1)
```

```python
y = df4.sold_out
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10)
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
gnb = GaussianNB()
```

```python
gnb.fit(X, y)
```

```python
y_pred = gnb.predict(X_test)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, y_pred))
```

```python
gnb.score(X_train, y_train)
```
