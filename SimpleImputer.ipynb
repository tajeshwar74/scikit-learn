```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({'miles': [20, 48, np.nan, 28, 43, 48]})
```

```python
df
```

```python
from sklearn.impute import SimpleImputer
```

```python
imp_mean = SimpleImputer(strategy='mean')
```

```python
imp_mean.fit_transform(df)
```

```python
imp_median = SimpleImputer(strategy='median')
```

```python
imp_median.fit_transform(df)
```

```python
imp_mode = SimpleImputer(strategy='most_frequent')
```

```python
imp_mode.fit_transform(df)
```

```python
imp_constant = SimpleImputer(strategy='constant', fill_value=13)
```

```python
imp_constant.fit_transform(df)
```

```python
#categorical data
names = pd.DataFrame({'name': ['billy', 'tommy', 'jackson', np.nan, 'chloe']})
```

```python
names
```

```python
imp_constant_cat = SimpleImputer(strategy='constant', fill_value='gekko')
```

```python
imp_constant_cat.fit_transform(names)
```

```python
imp_mean_marked = SimpleImputer(strategy='mean', add_indicator=True)
```

```python
imp_mean_marked.fit_transform(df)
```

```python
df = pd.DataFrame({'name': ['Ryan', 'Nolan', 'Walter', np.nan, 'Napolean'],
                   'miles': [50,62,np.nan,100,26]})
```

```python
df
```

```python
#
from sklearn.compose import make_column_transformer
```

```python
ct = make_column_transformer(
    (imp_constant_cat, ['name']),
    (imp_constant, ['miles']),
    remainder='drop'
)
```

```python
ct.set_output(transform='pandas')
```

```python
df_ct = ct.fit_transform(df)
```
